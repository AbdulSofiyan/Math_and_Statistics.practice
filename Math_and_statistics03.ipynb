```python
# Range : Range tells us the difference between the largest and smallest values(Range = Maximum − Minimum).
import numpy as np
data = [5, 10, 15, 20, 25]
print(np.max(data)-np.min(data))
```

```python
# finding the mean
data = [10,20,30,40]
mean = np.mean(data)
print(mean)
```

```python
# Variance : Variance tells us how far the values are spread from the mean.
data = [10,20,30]
print(np.var(data))
```

```python
#  Standard Deviation: Standard deviation is simply the square root of variance.

$$ SD = \sqrt{Variance} $$

For our example: SD = \sqrt66.67 ≈8.16
```

```python
# IQR : IQR measures the spread of the middle 50% of the data (IQR = Q3−Q1).
Q1 = 15
Q3 = 45
IQR = (Q3-Q1)
print(IQR)
```

```python
# Percentile : A percentile tells us the position of a value relative to the rest of the data.
# Quartiles : Quartiles divide sorted data into four parts.
data = [10, 20, 30, 40, 50]

print("Range:", np.max(data) - np.min(data))
print("Variance:", np.var(data))
print("Standard Deviation:", np.std(data))

print("Q1:", np.percentile(data, 25))
print("Q2:", np.percentile(data, 50))
print("Q3:", np.percentile(data, 75))
```
