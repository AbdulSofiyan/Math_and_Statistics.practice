```python
# mean - the averaage of all values
import numpy as np 
data =[10,20,30,40,50]
mean =  np.mean(data)
print(mean)
```

```python
# median - the median is the middle value after arranging the data in the data in asscending order
data =[10,20,30,40,50]
median =  np.median(data)
print(median)
```

```python
# if there even values ,then we have to do average of two middle values
data =[10,20,30,40]
median =  np.median(data)
print(median) # 20 + 30 = 50 (avg of 50 = 25)
```

```python
# mode - the mode is the value that occurs most frequently
from statistics import mode
data =[10,20,30,30,40]
print(mode(data))
```

```python
import pandas as pd

data = pd.Series([20000, 22000, 25000, 30000, 1000000])

print("Mean:", data.mean())
print("Median:", data.median())
print("Mode:", data.mode()[0])
```
