```python
--> 1. Probability: Probability tells us how likely an event is to happen.
P(Event)= Number of favorable outcomes/Total possible outcomes
probability is always between 0≤P(A)≤1

2. Sample Space: The sample space is the collection of all possible outcomes.
Coin : S={H,T}
Dice : S={1,2,3,4,5,6} 
3. Event: An event is the outcome we're interested in.
4. Complement of an Event: The complement means the event does NOT happen.
Formula:  P(A') = 1-P(A) 

5. Addition Rule: When we want the probability of A OR B:
For mutually exclusive events: P(A\ or\ B)=P(A)+P(B) 

6. Multiplication Rule: When we want the probability of A AND B for independent events:
formula: P(A and B)=P(A)×P(B)
```

```python
# A fair coin is tossed once. What is the probability of getting Tails?
# Probability = Favorable outcomes / Total outcomes

favorable = 1
total = 2

probability = favorable / total

print(probability)
print(probability * 100, "%")
```

```python
# Probability of getting an even number on a dice
favorable = 3
total = 6

probability = favorable / total

print(probability)
print(probability * 100, "%")
```

```python
--> 1. Conditional Probability:
        The probability of something happening when we already know that something else has happened.
Formula: P(A∣B)=P(B)P(A∩B)

Simple example:
Suppose a class has 100 students:
               60 students know Python
               40 students know Python and SQL

Question: If we already know a student knows Python, what is the probability that they also know SQL?

answer: Now our total group is no longer 100.

We only consider the 60 students who know Python.

So:  P(SQL|Python = frac{40}/{60} =0.667 

So the probability is approximately 66.7%.

The | symbol means "given that".

Therefore:

P(SQL | Python) = Probability of SQL given that the student knows Python.
```

```python
1. Independent Events: Two events are independent when one event does not affect the other.
2. Dependent Events: Two events are dependent when the first event affects the probability of the second event.
```
