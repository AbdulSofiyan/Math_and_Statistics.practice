```python
# 1. Limits:  What value does a function approach when x gets closer to a particular value.
# Question: Write Python code to calculate \(f(x)=x^2+3x\) for values of x increasingly close to 2
def f(x):
    return x**2 + 3*x
values = [1.9, 1.99, 1.999, 2.001, 2.01, 2.1]
for x in values:
    print(x, f(x))
```

```python
# 2.derivatives : A derivative tells us how quickly a value is changing.
# y → current value
# dy/dx → how quickly (y) changes when (x) changes
# Question: Using Python, define: f(x)=x^2 ,and calculate its derivative: f'(x)=2x, Find the derivative value when x = 5.
def derivative(x):
    return 2*x
value = [5]
for x in value:
    print(x,derivative(x))
```

```python
# 3. The Power Rule : This is the first derivative rule you should memorize.
# Question: Write Python code for: y=x^4, Using the Power Rule: dy/dx=4x^3, Calculate the derivative when x = 2.
def derivative(x):
    return 4 * x**3
value = [2]
for x in value:
    print(x, derivative(x))
```

```python
# 4. Constant Multiplier Rule : If a number is multiplied by a function, keep the number and differentiate the function.
# Question: Write Python code for: y=5x^3, Write Python code for its derivative: y′=15x^2, Calculate the derivative when x = 3.
def derivative(x):
    return 15*x**2
value = [3]
for x in value:
    print(x,derivative(x))
```

```python
# 5 Product Rule: Sometimes two functions are multiplied together.You cannot simply differentiate each part independently.
# The Product Rule is: (uv)′=u′v+uv′
# Question: Given: y=x^2(x+2) , Using the Product Rule: y′=2x(x+2)+x^2, Write Python code and calculate the derivative when x = 3.
def derivative(x):
    return 3*x**2 + 4*x

value = [3]

for x in value:
    print(x, derivative(x))
```

```python
# 7. Chain Rule: It is used when one function is inside another function (Derivative= outside derivative×inside derivative)
# Question: Given: y=(x^2+1)^3, Using the Chain Rule: y′=6x(x^2+1)^2, Write Python code and calculate the derivative when x = 2.
def derivative(x):
    return 6*x*(x**2+1)**2
value = [2]
for x in value:
    print(x,derivative(x))
```
